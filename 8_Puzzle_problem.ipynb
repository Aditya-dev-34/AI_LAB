```python
def get_moves(state):

    moves = []
    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    # Up
    if row > 0:
        new_state = state.copy()
        new_state[zero], new_state[zero - 3] = new_state[zero - 3], new_state[zero]
        moves.append(new_state)

    # Down
    if row < 2:
        new_state = state.copy()
        new_state[zero], new_state[zero + 3] = new_state[zero + 3], new_state[zero]
        moves.append(new_state)

    # Left
    if col > 0:
        new_state = state.copy()
        new_state[zero], new_state[zero - 1] = new_state[zero - 1], new_state[zero]
        moves.append(new_state)

    # Right
    if col < 2:
        new_state = state.copy()
        new_state[zero], new_state[zero + 1] = new_state[zero + 1], new_state[zero]
        moves.append(new_state)

    return moves


def dls(state, goal, limit, path, visited):
    if state == goal:
        return path

    if limit == 0:
        return "CUTOFF"

    cutoff_occurred = False

    for next_state in get_moves(state):

        state_tuple = tuple(next_state)

        if state_tuple in visited:
            continue

        visited.add(state_tuple)

        result = dls(
            next_state,
            goal,
            limit - 1,
            path + [next_state],
            visited
        )

        visited.remove(state_tuple)

        if result == "CUTOFF":
            cutoff_occurred = True

        elif result != "FAILURE":
            return result

    if cutoff_occurred:
        return "CUTOFF"
    else:
        return "FAILURE"


print("Enter initial state (use 0 for blank):")
initial = list(map(int, input().split()))

print("Enter goal state (use 0 for blank):")
goal = list(map(int, input().split()))

limit = int(input("Enter depth limit: "))

visited = {tuple(initial)}

result = dls(
    initial,
    goal,
    limit,
    [initial],
    visited
)

if result == "CUTOFF":
    print("\nCUTOFF: Goal may exist beyond the depth limit.")

elif result == "FAILURE":
    print("\nFAILURE: No solution found within the given depth limit.")

else:
    print("\nSolution found!")
    print("Number of moves:", len(result) - 1)

    print("\nPath:")

    for i, state in enumerate(result):
        print("Step", i)
        print(state[0], state[1], state[2])
        print(state[3], state[4], state[5])
        print(state[6], state[7], state[8])
        print()
```
