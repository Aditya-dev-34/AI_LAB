```python

def unify(a, b, theta):
    if a == b:
        return theta

    if a.islower():
        theta[a] = b
        return theta

    if b.islower():
        theta[b] = a
        return theta

    return None


e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

a = e1.replace("(", " ").replace(")", " ").replace(",", " ").split()
b = e2.replace("(", " ").replace(")", " ").replace(",", " ").split()

if len(a) != len(b) or a[0] != b[0]:
    print("Unification Failed")

else:
    theta = {}
    success = True

    for x, y in zip(a[1:], b[1:]):
        result = unify(x, y, theta)

        if result is None:
            success = False
            break

    if success:
        print("Unification Successful")
        print("Substitution:", theta)
    else:
        print("Unification Failed")
```
