```python
from itertools import product

def implies(a, b):
    return (not a) or b

def tt_entails():
    symbols = ["P", "Q", "R"]
    entails = True
    models = 0

    print("P     Q     R     | P->Q  Q->R  | KB    | Query R | Result (KB->R)")
    print("-----------------------------------------------------------------------")

    for values in product([True, False], repeat=3):
        model = dict(zip(symbols, values))

        P = model["P"]
        Q = model["Q"]
        R = model["R"]

        p_implies_q = implies(P, Q)
        q_implies_r = implies(Q, R)

        KB = P and p_implies_q and q_implies_r
        result = implies(KB, R)

        str_P = "True" if P else "False"
        str_Q = "True" if Q else "False"
        str_R = "True" if R else "False"
        str_p_implies_q = "True" if p_implies_q else "False"
        str_q_implies_r = "True" if q_implies_r else "False"
        str_KB = "True" if KB else "False"
        str_result = "True" if result else "False"

        print(
            f"{str_P:<5} {str_Q:<5} {str_R:<5} | "
            f"{str_p_implies_q:<5} {str_q_implies_r:<5} | "
            f"{str_KB:<5} | "
            f"{str_R:<7} | "
            f"{str_result:<5}"
        )

        if KB:
            models += 1
            if not R:
                entails = False

    print("\nModels satisfying KB:", models)

    if entails:
        print("KB entails R: TRUE")
    else:
        print("KB does not entail R: FALSE")

tt_entails()
```

```python
from itertools import product

def implies(a, b):
    return (not a) or b

def tt_entails_q():
    symbols = ["P", "Q"]
    entails = True
    models = 0

    print("P     Q     | P->Q (KB) | Query Q | Result (KB->Q)")
    print("--------------------------------------------------")

    for values in product([True, False], repeat=2):
        model = dict(zip(symbols, values))

        P = model["P"]
        Q = model["Q"]

        KB = implies(P, Q)
        result = implies(KB, Q)

        str_P = "True" if P else "False"
        str_Q = "True" if Q else "False"
        str_KB = "True" if KB else "False"
        str_result = "True" if result else "False"

        print(f"{str_P:<5} {str_Q:<5} | {str_KB:<9} | {str_Q:<7} | {str_result:<5}")

        if KB:
            models += 1
            if not Q:
                entails = False

    print("\nModels satisfying KB:", models)

    if entails:
        print("KB entails Q: TRUE")
    else:
        print("KB does not entail Q: FALSE")

tt_entails_q()
```
