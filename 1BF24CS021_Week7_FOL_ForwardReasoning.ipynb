```python
KB = {
    "American(Robert)",
    "Enemy(A,America)",
    "Owns(A,T1)",
    "Missile(T1)"
}

def forward_chaining(KB):
    while True:
        new = set()

        if "Missile(T1)" in KB:
            new.add("Weapon(T1)")

        if "Missile(T1)" in KB and "Owns(A,T1)" in KB:
            new.add("Sells(Robert,T1,A)")

        if "Enemy(A,America)" in KB:
            new.add("Hostile(A)")

        if {"American(Robert)", "Weapon(T1)",
            "Sells(Robert,T1,A)", "Hostile(A)"}.issubset(KB):
            new.add("Criminal(Robert)")

        if new.issubset(KB):
            break

        KB.update(new)
        print("New Facts:", new)

    return "Criminal(Robert)" in KB


print("Initial Knowledge Base:", KB)
result = forward_chaining(KB)

print("\nQuery: Criminal(Robert)")
print("Result:", result)

if result:
    print("Robert is Criminal.")
else:
    print("Robert is not proved Criminal.")
```
