```python
from heapq import heappush, heappop


def misplaced_tiles(state, goal):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count


def get_neighbors(state):
    neighbors = []
    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    if row > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 3] = (
            new_state[zero - 3],
            new_state[zero],
        )
        neighbors.append(tuple(new_state))

    if row < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 3] = (
            new_state[zero + 3],
            new_state[zero],
        )
        neighbors.append(tuple(new_state))

    if col > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 1] = (
            new_state[zero - 1],
            new_state[zero],
        )
        neighbors.append(tuple(new_state))

    if col < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 1] = (
            new_state[zero + 1],
            new_state[zero],
        )
        neighbors.append(tuple(new_state))

    return neighbors


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def a_star(initial, goal):
    open_list = []
    closed = set()
    g_cost = {}
    parent = {}

    g_cost[initial] = 0
    parent[initial] = None
    h = misplaced_tiles(initial, goal)
    f = g_cost[initial] + h
    heappush(open_list, (f, 0, initial))

    while open_list:
        f, g, current = heappop(open_list)

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            path.reverse()
            return path

        if current in closed:
            continue

        closed.add(current)

        for next_state in get_neighbors(current):
            if next_state in closed:
                continue

            new_g = g + 1

            if next_state not in g_cost or new_g < g_cost[next_state]:
                g_cost[next_state] = new_g
                parent[next_state] = current
                h = misplaced_tiles(next_state, goal)
                f = new_g + h
                heappush(open_list, (f, new_g, next_state))

    return None


print("8-PUZZLE A* SEARCH")
print("Use 0 for the blank space")
print()

print("Enter Initial State:")
initial = tuple(map(int, input().split()))

print("Enter Goal State:")
goal = tuple(map(int, input().split()))

if len(initial) != 9 or len(goal) != 9:
    print("Error: Enter exactly 9 numbers.")
else:
    solution = a_star(initial, goal)
    if solution is not None:
        print("\nSOLUTION FOUND")
        print("------------------------")
        print("Number of moves:", len(solution) - 1)
        print()
        for step, state in enumerate(solution):
            g = step
            h = misplaced_tiles(state, goal)
            f = g + h
            print("Step:", step)
            print_state(state)
            print("g(n) =", g)
            print("h(n) =", h)
            print("f(n) =", f)
            print("------------------------")
    else:
        print("No solution exists.")
```
