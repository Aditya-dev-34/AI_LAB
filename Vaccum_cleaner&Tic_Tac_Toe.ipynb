### Generalized Vacuum Cleaner Agent
To automate this code for an arbitrary number of rooms, we can represent the rooms as a list. The vacuum can then iterate through this list, cleaning each dirty room it encounters.

```python
import random

def run_vacuum_agent(num_rooms):
    rooms = []
    print(f"\nInitial State Input ({num_rooms} Rooms):")
    for i in range(num_rooms):
        while True:
            try:
                state = int(input(f"Enter state for Room {i+1} (0 for clean, 1 for dirty): "))
                if state in [0, 1]:
                    rooms.append(state)
                    break
                else:
                    print("Invalid input. Please enter 0 or 1.")
            except ValueError:
                print("Invalid input. Please enter an integer (0 or 1).")

    print(f"\nInitial State ({num_rooms} Rooms):")
    for i, state in enumerate(rooms):
        print(f"Room {i+1} = {'Dirty' if state == 1 else 'Clean'}")

    score = 0
    print("\nVacuum Cleaning Process:")
    for i in range(num_rooms):
        print(f"\nVacuum is at Room {i+1}")
        if rooms[i] == 1:
            print(f"Room {i+1} is Dirty -> Suck")
            rooms[i] = 0  # Clean the room
            score += 1
        else:
            print(f"Room {i+1} is Clean")

    print("\nFinal State:")
    for i, state in enumerate(rooms):
        print(f"Room {i+1} = {'Dirty' if state == 1 else 'Clean'}")

    print("\nPerformance Measurement (cleaned rooms):", score)

# --- Demo for 4 rooms ---
print("\n--- Running for 4 Rooms (User Input) ---")
run_vacuum_agent(4)

# --- Demo for 2 rooms ---
print("\n--- Running for 2 Rooms (User Input) ---")
run_vacuum_agent(2)
```
