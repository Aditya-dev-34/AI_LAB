```python
from heapq import heappush, heappop


def manhattan_distance(state, goal):

    distance = 0

    for tile in range(1, 9):

        current_position = state.index(tile)

        current_row = current_position // 3
        current_col = current_position % 3


        goal_position = goal.index(tile)

        goal_row = goal_position // 3
        goal_col = goal_position % 3


        distance += abs(current_row - goal_row) + \
                    abs(current_col - goal_col)

    return distance

def get_neighbors(state):

    neighbors = []

    zero = state.index(0)

    row = zero // 3
    col = zero % 3


    if row > 0:

        new_state = list(state)

        new_state[zero], new_state[zero - 3] = \
            new_state[zero - 3], new_state[zero]

        neighbors.append(tuple(new_state))


    if row < 2:

        new_state = list(state)

        new_state[zero], new_state[zero + 3] = \
            new_state[zero + 3], new_state[zero]

        neighbors.append(tuple(new_state))

    if col > 0:

        new_state = list(state)

        new_state[zero], new_state[zero - 1] = \
            new_state[zero - 1], new_state[zero]

        neighbors.append(tuple(new_state))

    if col < 2:

        new_state = list(state)

        new_state[zero], new_state[zero + 1] = \
            new_state[zero + 1], new_state[zero]

        neighbors.append(tuple(new_state))

    return neighbors



def print_state(state):

    for i in range(0, 9, 3):

        print(state[i], state[i + 1], state[i + 2])

    print()



def a_star(initial, goal):

    # OPEN list
    # Stores: (f, g, state)
    open_list = []

    # CLOSED set
    closed = set()

    # Store g(n) values
    g_cost = {}

    # Store parent of each state
    parent = {}


    g_cost[initial] = 0

    parent[initial] = None

    # h(n) = Manhattan Distance
    h = manhattan_distance(initial, goal)

    # f(n) = g(n) + h(n)
    f = g_cost[initial] + h

    heappush(open_list, (f, 0, initial))



    while open_list:

        # Select state with minimum f(n)
        f, g, current = heappop(open_list)



        if current == goal:

            path = []

            while current is not None:

                path.append(current)

                current = parent[current]

            path.reverse()

            return path

        # Skip if already explored
        if current in closed:

            continue

        # Add current state to CLOSED
        closed.add(current)



        for next_state in get_neighbors(current):

            if next_state in closed:

                continue

            # g(next) = g(current) + 1
            new_g = g + 1

            # If new state or better path found
            if next_state not in g_cost or \
               new_g < g_cost[next_state]:

                # Store new g(n)
                g_cost[next_state] = new_g

                # Store parent
                parent[next_state] = current

                # Calculate h(n)
                h = manhattan_distance(next_state, goal)

                # Calculate f(n)
                f = new_g + h

                # Add to OPEN
                heappush(
                    open_list,
                    (f, new_g, next_state)
                )

    return None




print("======================================")
print("   8-PUZZLE A* - MANHATTAN DISTANCE")
print("======================================")

print("\nUse 0 for the blank space.")


print("\nEnter Initial State:")
initial = tuple(map(int, input().split()))


print("\nEnter Goal State:")
goal = tuple(map(int, input().split()))




if len(initial) != 9 or len(goal) != 9:

    print("\nError: Please enter exactly 9 numbers.")

elif set(initial) != set(range(9)) or \
     set(goal) != set(range(9)):

    print("\nError: Use numbers 0 to 8 exactly once.")

else:


    solution = a_star(initial, goal)


    if solution is not None:

        print("\n======================================")
        print("          SOLUTION FOUND")
        print("======================================")

        print("Number of moves:", len(solution) - 1)

        print()

        for step, state in enumerate(solution):

            # g(n) = depth
            g = step

            # h(n) = Manhattan Distance
            h = manhattan_distance(state, goal)

            # f(n) = g(n) + h(n)
            f = g + h

            print("Step:", step)

            print_state(state)

            print("g(n) =", g)
            print("h(n) =", h)
            print("f(n) =", f)

            print("--------------------------------------")

    else:

        print("\nNo solution exists.")
```
